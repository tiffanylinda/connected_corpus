### Part 2: Using the Python libraries to scrape a webpage

```python
def getHTMLPage(url):
    """Given a url, get the HTML page content"""
    
    response = requests.get(url)
    if response.status_code == 200:
        return response.text
    else:
        print("Failure resaon:", response.reason)
        return
```

```python
url = "https://theconnectedcorpus.pubpub.org/pub/1mdeh2xj/release/4"
htmlPage = getHTMLPage(url)
print(htmlPage)
```

```python
# Creating an HTML page parser/reader
domTree = BS(htmlPage, 'html.parser')
type(domTree)
```

```python
# Finding all the annotations, they were all under the class "discussion-list"
comments = domTree.find(class_ = "discussion-list")

# The method .prettify allows us to see the structure of the HTML page better . 
print(comments.prettify())
```

```python
# Iterating through the comments data, and pulling out the annotations as strings 
# All the annoations were under the <p> tag in the HTML
for el in comments.find_all("p"):
    print(el.text)
```

```python
# My Professor helped with the following parts below. She said that the annotation data was most likely stored in the script
scripts = domTree.find_all("script")
len(scripts)
```

```python
#Iterates through the scripts
for i, s in enumerate(scripts):
    print(i, str(s)[:100])
```

```python
script = scripts[4]
data = script['data-json']
len(data)
```

```python
# Checking what type of data structure pubdata is
import json
pubdata = json.loads(data)
type(pubdata)
```

```python
# We can see what keys/categories each of the data is listed under
pubdata['pubData'].keys()
```

```python
# Extracting links in tuples
links_list = []
for el in comments.find_all("p"):
    links = tuple(a['href'] for a in el.find_all("a"))
    links_list.append(links)

print(links_list)
```

Run the code below if you just want to count the number of links in a webpage

```python
# Extracting links in tuples
links_list = []
count = 0
for el in comments.find_all("p"):
    links = tuple(a['href'] for a in el.find_all("a"))
    count += len(links)
    links_list.append(links)

print(links_list)
print(count)
```
